```python
import numpy as np
import matplotlib.pyplot as plt


dma_data = np.loadtxt('./data/PS1_prior_post.csv', skiprows=1, dtype=float, delimiter=',')



norm_window = (15, 25)


diameter_mp = dma_data[:, 0]
counts_prior = dma_data[:, 1]
counts_post = dma_data[:, 3]
in_window = (diameter_mp >= norm_window[0]) & (diameter_mp <= norm_window[1])



plt.rcParams.update({'font.size': 7})
plt.rcParams.update({"font.family":"sans-serif"})
plt.rcParams.update({"font.sans-serif":"Nimbus Sans"})
plt.rcParams.update({"pdf.fonttype":42})

fig, ax = plt.subplots(figsize=(3.5, 2.625), dpi=600)

ax.plot(diameter_mp, counts_prior / counts_prior[in_window].max(), color='black', linewidth=2, label='Prior to experiment')
ax.plot(diameter_mp, counts_post / counts_post[in_window].max(), color='red', linewidth=2, label='Post experiment')



ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.set_xlim(5, 40)
ax.set_ylim(0, 2.3)
ax.set_xlabel('PS1 particle size (nm)')
ax.set_ylabel('Normalized data (a.u.)')
ax.legend(fancybox=False, frameon=False)
plt.tight_layout()
plt.savefig('./figs/PS1_prior_post.png')
plt.savefig('./figs/PS1_prior_post.svg', format='svg', bbox_inches='tight', dpi=600, transparent=True)
plt.show()
```
